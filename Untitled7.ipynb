```python
# Make a working copy
cleaned_df = df.copy()

# 1. Remove extra spaces from column names
cleaned_df.columns = cleaned_df.columns.str.strip()

# 2. Remove unwanted index columns such as Unnamed: 0
unnamed_cols = [
    col for col in cleaned_df.columns
    if col.lower().startswith("unnamed:")
]

cleaned_df.drop(columns=unnamed_cols, inplace=True)

print("Removed unnecessary columns:", unnamed_cols)

# 3. Remove duplicate rows
duplicates_before = cleaned_df.duplicated().sum()
cleaned_df.drop_duplicates(inplace=True)

print("Duplicate rows removed:", duplicates_before)

# 4. Remove leading/trailing spaces from text columns
for col in cleaned_df.select_dtypes(include="object").columns:
    cleaned_df[col] = cleaned_df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# 5. Replace empty strings with missing values
cleaned_df.replace(r"^\s*$", np.nan, regex=True, inplace=True)

# 6. Convert numeric columns when they exist
numeric_columns = [
    "magnitude", "depth", "latitude", "longitude",
    "sig", "cdi", "mmi", "tsunami", "year"
]

for col in numeric_columns:
    if col in cleaned_df.columns:
        cleaned_df[col] = pd.to_numeric(
            cleaned_df[col], errors="coerce"
        )

# 7. Convert date/time column when present
if "date_time" in cleaned_df.columns:
    cleaned_df["date_time"] = pd.to_datetime(
        cleaned_df["date_time"],
        errors="coerce",
        dayfirst=True
    )

# 8. Standardize text categories
for col in ["alert", "country", "status", "location"]:
    if col in cleaned_df.columns:
        cleaned_df[col] = cleaned_df[col].replace(
            r"^\s*$", np.nan, regex=True
        )

print("Basic cleaning completed!")
```

```python
# Validate latitude: -90 to 90
if "latitude" in cleaned_df.columns:
    invalid_lat = ~cleaned_df["latitude"].between(-90, 90)
    cleaned_df.loc[invalid_lat, "latitude"] = np.nan

# Validate longitude: -180 to 180
if "longitude" in cleaned_df.columns:
    invalid_lon = ~cleaned_df["longitude"].between(-180, 180)
    cleaned_df.loc[invalid_lon, "longitude"] = np.nan

print("Geographic validation completed.")
```

```python
# Fill missing categorical values with "Unknown"
categorical_columns = cleaned_df.select_dtypes(
    include=["object", "category"]
).columns

for col in categorical_columns:
    cleaned_df[col] = cleaned_df[col].fillna("Unknown")

# Keep missing numeric values as NaN for accurate analysis
print("Missing values after cleaning:")
print(cleaned_df.isnull().sum())
```

```python
print("=" * 45)
print("       DATA CLEANING REPORT")
print("=" * 45)

print("Original rows:", len(df))
print("Cleaned rows:", len(cleaned_df))
print("Rows removed:", len(df) - len(cleaned_df))

print("\nOriginal columns:", len(df.columns))
print("Cleaned columns:", len(cleaned_df.columns))

print("\nRemaining duplicate rows:",
      cleaned_df.duplicated().sum())

print("\nMissing values per column:")
print(cleaned_df.isnull().sum())

print("\nCleaned dataset preview:")
display(cleaned_df.head())
```

```python
sentiment_counts = df["Sentiment"].value_counts().reset_index()
sentiment_counts.columns = ["Sentiment", "Count"]

# Create interactive pie chart
fig = px.pie(
    sentiment_counts,
    names="Sentiment",
    values="Count",
    title="Distribution of Posts by Sentiment",
    hover_data=["Count"],
    hole=0  # Set to 0.4 for a donut chart
)

# Display percentages and category names
fig.update_traces(
    textinfo="label+percent",
    hovertemplate=(
        "Sentiment: %{label}<br>"
        "Number of Posts: %{value}<br>"
        "Percentage: %{percent}<extra></extra>"
    )
)

# Customize graph
fig.update_layout(
    template="plotly_white"
)

# Display chart
fig.show()
```

```python
# Select numerical columns
numeric_df = df.select_dtypes(include="number")
# Calculate correlation matrix
corr_matrix = numeric_df.corr()
# Create interactive heatmap
fig = px.imshow(
    corr_matrix,
    text_auto=".2f",
    color_continuous_scale="RdBu_r",
    zmin=-1,
    zmax=1,
    aspect="auto",
    title="Correlation Heatmap of Numerical Features",
    labels=dict(
        x="Features",
        y="Features",
        color="Correlation"
    )
)
# Customize graph
fig.update_layout(
    template="plotly_white",
    xaxis_title="Features",
    yaxis_title="Features"
)
# Display graph
fig.show()
```

```python
# Convert Timestamp to datetime
df["Timestamp"] = pd.to_datetime(df["Timestamp"], errors="coerce")
# Remove rows with missing timestamps or likes
df = df.dropna(subset=["Timestamp", "Likes"])
# Sort records by time
df = df.sort_values("Timestamp")
# Group data by date and calculate average likes
df["Date"] = df["Timestamp"].dt.date
daily_likes = (
    df.groupby("Date", as_index=False)["Likes"]
    .mean()
)
# Create interactive line chart
fig = px.line(
    daily_likes,
    x="Date",
    y="Likes",
    markers=True,
    title="Average Likes Over Time",
    labels={
        "Date": "Date",
        "Likes": "Average Number of Likes"
    },
    hover_data={"Likes": ":.2f"}
)
# Customize graph
fig.update_layout(
    template="plotly_white",
    xaxis_title="Date",
    yaxis_title="Average Likes",
    hovermode="x unified"
)
# Display graph
fig.show()
```
