```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("sentimentdataset_github.csv")

# Remove unnecessary index columns
df = df.drop(columns=['Unnamed: 0.1', 'Unnamed: 0'], errors='ignore')

# Display basic information
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x='Sentiment',
    y='Likes',
    hue='Sentiment',
    palette='Set2',
    legend=False
)

plt.title('Likes Distribution by Sentiment (Box Plot)')
plt.xlabel('Sentiment')
plt.ylabel('Likes')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.violinplot(
    data=df,
    x='Sentiment',
    y='Likes',
    hue='Sentiment',
    palette='Set2',
    legend=False
)

plt.title('Likes Distribution by Sentiment (Violin Plot)')
plt.xlabel('Sentiment')
plt.ylabel('Likes')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='Likes',
    bins=20,
    kde=True,
    color='skyblue'
)

plt.title('Distribution of Likes(Histogram)')
plt.xlabel('Likes')
plt.ylabel('Number of Posts')

plt.tight_layout()
plt.show()
```

```python
num_cols = ['Retweets', 'Likes', 'Month', 'Day', 'Hour']

corr = df[num_cols].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    square=True,
    fmt='.2f'
)

plt.title('Correlation Matrix of Sentiment Dataset')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='Sentiment',
    hue='Sentiment',
    palette='Set2',
    legend=False
)

plt.title('Number of Posts by Sentiment')
plt.xlabel('Sentiment')
plt.ylabel('Number of Posts')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
